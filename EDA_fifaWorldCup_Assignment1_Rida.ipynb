Theoretical: why does correct data typing matter? (Think about calculations, sorting,  dates and feeding the data to a model.)

1. Calculations: Numeric operations require numeric types. String '5' + '5' = '55',
   not 10. Mean, sum, etc. won't work with text.

2. Sorting: Text sorting is alphabetical (100 < 2), while numeric sorting is
   by value (2 < 100). This affects rankings and ordering.

3. Dates: Date objects enable time-based analysis (e.g., grouping by year,
   calculating days between dates). Text dates can't do this.

4. Memory: Numeric types use less memory than object types.

5. Model compatibility: Machine learning models require numeric inputs.

6. Data quality: Correct types help identify outliers and anomalies.

Q8: Missing Data Strategy & Impossible Values (3 marks)
Decide how to handle the missing values, and justify every choice. Apply your strategy in code.
1. For the numeric columns (Age, Goals, Market Value, Assists, Matches_Played):  fill the missing values with the mean or the median? Compare the two and justify  your choice based on whether the column looks skewed.


```python


# FIX GOALS
print("\n Fixing Goals column...")
df['Goals'] = df['Goals'].replace(['N/A', '-', '?', 'N/A', 'na'], np.nan)
df['Goals'] = pd.to_numeric(df['Goals'], errors='coerce')
df['Goals'] = df['Goals'].fillna(df['Goals'].mean())
print(f"   Goals fixed! Mean: {df['Goals'].mean():.2f}")

# FIX MARKET VALUE
print("\n Fixing Market Value column...")
# Replace text placeholders
df['Market Value'] = df['Market Value'].replace(['Unknown', '-', 'N/A'], np.nan)
# Remove € and M
df['Market Value'] = df['Market Value'].astype(str).str.replace('€', '')
df['Market Value'] = df['Market Value'].astype(str).str.replace('M', '')
df['Market Value'] = df['Market Value'].astype(str).str.strip()
# Convert to number
df['Market Value'] = pd.to_numeric(df['Market Value'], errors='coerce')
# Fill missing with median (because it's skewed)
df['Market Value'] = df['Market Value'].fillna(df['Market Value'].median())
print(f"   Market Value fixed! Median: {df['Market Value'].median():.2f}")

# FIX ASSISTS
print("\nFixing Assists column...")
df['Assists'] = pd.to_numeric(df['Assists'], errors='coerce')
df['Assists'] = df['Assists'].fillna(df['Assists'].mean())
print(f"   Assists fixed! Mean: {df['Assists'].mean():.2f}")

# FIX MATCHES_PLAYED
print("\n Fixing Matches_Played column...")
df['Matches_Played'] = pd.to_numeric(df['Matches_Played'], errors='coerce')
df['Matches_Played'] = df['Matches_Played'].fillna(df['Matches_Played'].mean())
print(f"   Matches_Played fixed! Mean: {df['Matches_Played'].mean():.2f}")

# FIX AGE
print("\nFixing Age column...")
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(f"   Age fixed! Mean: {df['Age'].mean():.2f}")

# CHECK RESULTS
print("\n Final check for missing values:")
print("-"*40)
missing = df[['Age', 'Goals', 'Market Value', 'Assists', 'Matches_Played']].isnull().sum()
for col in missing.index:
    if missing[col] == 0:
        print(f" {col}: No missing values")
    else:
        print(f"  {col}: {missing[col]} missing")
```

2. For Club (a text column with missing values): fill with the mode, or use another  approach? Justify.

```python

club_missing_before = df['Club'].isnull().sum()
print(f"Missing before: {club_missing_before}")

if club_missing_before > 0:
    # Get mode (most frequent)
    mode_club = df['Club'].mode()[0]
    print(f"Mode (most frequent club): '{mode_club}'")

    # Fill with mode
    df['Club'] = df['Club'].fillna(mode_club)

    club_missing_after = df['Club'].isnull().sum()
    print(f"Missing after: {club_missing_after}")
    print(f" Filled with mode - '{mode_club}'")
    print(f"   Justification: Mode is the most common value, best guess for missing")
    print(f"   Alternative: Could use 'Unknown' but mode preserves data distribution")
```

4. After converting Market Value, compare the average market value of players  older than 30 with those aged 30 and under (masking + .mean()). What does the  difference suggest?


```python
# First convert Market Value to numeric
df['Market Value'] = df['Market Value'].replace(['Unknown', '-', 'N/A'], np.nan)
df['Market Value'] = df['Market Value'].astype(str).str.replace('€', '')
df['Market Value'] = df['Market Value'].astype(str).str.replace('M', '')
df['Market Value'] = pd.to_numeric(df['Market Value'], errors='coerce')

# Calculate averages using Boolean masking
avg_mv_older = df[df['Age'] > 30]['Market Value'].mean()
avg_mv_younger = df[df['Age'] <= 30]['Market Value'].mean()

# Count players
older_count = len(df[df['Age'] > 30])
younger_count = len(df[df['Age'] <= 30])

print(f"   Average market value (older >30): €{avg_mv_older:.1f}M")
print(f"   Average market value (younger ≤30): €{avg_mv_younger:.1f}M")
print(f"   Difference: €{avg_mv_younger - avg_mv_older:+.1f}M")
print(f"   Number of older players: {older_count}")
print(f"   Number of younger players: {younger_count}")

# Insight
if avg_mv_younger > avg_mv_older:
    print(f"\nINSIGHT: Younger players (≤30) have €{avg_mv_younger - avg_mv_older:.1f}M higher market value.")
    print(f"   This suggests clubs value youth, potential, and future resale value.")
else:
    print(f"\nINSIGHT: Older players (>30) have €{avg_mv_older - avg_mv_younger:.1f}M higher market value.")
    print(f"   This suggests experience and proven track record are valued.")
```
